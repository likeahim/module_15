```python
import numpy as np

# Wektor od 1 do 10
vec1 = np.arange(1, 11)

# Wektor 10 jedynek (ładniejszy sposób)
vec2 = np.ones(10)

# Dodawanie wektorów
result = vec1 + vec2

print("Wektor 1:", vec1)
print("Wektor 2:", vec2)
print("Suma wektorów:", result)
```

```python
# Macierz 5x5 z liczbami od 1 do 25
matrix = np.arange(1, 26).reshape(5, 5)

# Średnie dla wierszy (axis=1)
row_means = matrix.mean(axis=1)

# Średnie dla kolumn (axis=0)
col_means = matrix.mean(axis=0)

print("Macierz:\n", matrix)
print("Średnie wartości dla każdego wiersza:", row_means)
print("Średnie wartości dla każdej kolumny:", col_means)
```
